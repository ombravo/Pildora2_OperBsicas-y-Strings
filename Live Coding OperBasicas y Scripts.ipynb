Operadores artiméticos básicos

```python
print(2+4)
print(4-2)
print(2*4)
print(9/2)
print(2**4) #Potencia- 2 elevado a 4
```

Diseccionando la división

```python
print(17//5) #Division entera- 17 dividido entre 5, sin decimales
print(17%5) #Modulo- resto de la division 17 entre 5, el cociente entero es 3 (5 X 3 = 15) y sobra 23 como resto
```

```python
print(17/0) #Division entre cero, error
```

Operadores de Comparación y Lógicos

Igualdad y Desigualdad: Comparar si dos valores son iguales o diferentes

```python
print(5==5) #True, es una comparacion, no una asignacion
print(5!=3) #True, es una comparacion, no una asignacion
```

Menor y mayor

```python
print(3<5)
print(5>3)
print(3<=5)
print (3>5)
```

Operadores Lógicos

Cadenas (Strings): El Texto como Secuencia

```python
print("Cuesta " + 3 + "euros") #Error, no se puede concatenar un string con un numero
```

```python
print("Cuesta " + str(3) + " euros") #Correcto, se convierte el numero a string
```

Slicing: Extracción de Subcadenas

```python
texto = "Agua pasada no mueve molino"
print(texto[0:4]) #Agua
print(texto[12:]) #no mueve molino
print(texto[::-1]) #olinom evuem on sap auguA
```

Formateo Moderno: f-strings

```python
nombre = "Ana"
edad = 25

mensaje_antes = "Me llamo " + nombre + " y tengo " + str(edad) + " años"
print(mensaje_antes)


mensaje_fstring = f"Me llamo {nombre} y tengo {edad} años"
print(mensaje_fstring)
```

Limpieza y Reemplazo de Datos

```python
#.strip

web = " www.empresa.com " #dirección con espacios antes y después
web_limpia = web.strip() #
print(web_limpia)  # Output: "www.empresa.com"
```

```python
#.replace

saludo = "Estimado nombre"
saludo_personalizado = saludo.replace("nombre", "Juan")

print(saludo_personalizado)  # Output: "Estimado Juan"[cite: 3]
print(saludo)                # Sigue intacto: "Estimado nombre"[cite: 3]
```

Fragmentación de Textos

```python
#.split
texto = "python, guia, curso, tutorial"
lista = texto.split(", ")

print(lista)  # Salida: ['python', 'guia', 'curso', 'tutorial']
```
